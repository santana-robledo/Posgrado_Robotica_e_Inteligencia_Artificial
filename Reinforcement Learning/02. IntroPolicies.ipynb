## Reto 1
Modifica el episodio para la recompensa quede así:
*  si la recompensa es 0, cambiarla a -1
*  si la recompensa es 1, cambiarla a 0
* En cada paso de tiempo imprime la recompensa
Imprime la recompensa total al final del episodio

```python
env = gym.make('FrozenLake-v1', desc=None, map_name="4x4", is_slippery=False, render_mode = "human")
observation, _ = env.reset() 
done = False
total_reward = 0
while not done:
    #Escoger una acción random
    action = env.action_space.sample() 
    #Usar la acción en el ambiente
    observation, reward, done, truncated, info = env.step(action)
    # cambiar recompensa a -1 o 0 respectivamente
    if reward ==0:
        reward= -1
    else:
        reward=0

    #imprimir recompensa
    print(reward)
    #acumular recompensa total
    total_reward+=reward

#Imprimir recompensa total
print(f'total reward: {total_reward}')
#Cerrar el ambiente    
env.close()
```
