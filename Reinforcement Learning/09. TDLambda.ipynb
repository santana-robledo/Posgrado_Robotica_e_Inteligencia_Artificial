```python
import time 
import gymnasium as gym
from tqdm import tqdm  # tqdm es una libreria que muestra barras de progreso
import random
import numpy as np 
import matplotlib.pyplot as plt

print(gym.__version__)
```

```python
class QLambda:
    """ Q(Lambda) algorithm """
    def __init__(self, epsilon, epsilon_min, epsilon_decay, alpha, gamma, lam, n_states, n_action):
        self.Q_table = np.zeros([n_states, n_action])  # Inicializar Q(s,a) table
        self.traces = np.zeros([n_states, n_action])  # Initializar tabla de rasgos
        self.n_action = n_action  # numero de acciones
        self.alpha = alpha  # tasa de aprendizaje
        self.gamma = gamma  # factor de descuento
        self.lam = lam # descuento de rasgo
        self.epsilon = epsilon  # Parametros epsilon-greedy
        self.epsilon_min = epsilon_min
        self.epsilon_decay = epsilon_decay

    def take_action(self, state):  # epsilon-greedy seleccion de accion
        if np.random.random() < self.epsilon:
            action = np.random.randint(self.n_action)
        else:
            action = np.argmax(self.Q_table[state])
        return action

    def decay_epsilon(self):
        self.epsilon = max(self.epsilon_min, self.epsilon*self.epsilon_decay)

    def best_action(self, state):  # para imprimir la estrategia
        Q_max = np.max(self.Q_table[state])
        a = [0 for _ in range(self.n_action)]
        for i in range(self.n_action):
            if self.Q_table[state, i] == Q_max:
                a[i] = 1
        return a

    def update(self, s0, a0, r, s1):
        self.traces[s0, a0] = 1
        td_error = r + self.gamma * self.Q_table[s1].max() - self.Q_table[s0, a0]
        self.Q_table += self.alpha * td_error * self.traces
        self.traces *= self.gamma * self.lam
```

```python
env = gym.make('CliffWalking-v1')

n_states = env.observation_space.n
n_actions = env.action_space.n
np.random.seed(0)
epsilon = 1
epsilon_min = 0.01
epsilon_decay = 0.995
alpha = 0.5
gamma = 0.99
lam = 0.3
agent = QLambda(epsilon, epsilon_min, epsilon_decay, alpha, gamma, lam, n_states, n_actions)
num_episodes = 2000  # Numero de secuencia que correra el agente en el ambiente

return_list = []  # Guardar el retorno de cada secuencia
for i in range(10):  # Mostrar 10 barras de progreso
    # Barra de progreso de la funcion tdqm
    with tqdm(total=int(num_episodes / 10), desc='Iteration %d' % i) as pbar:
        for i_episode in range(int(num_episodes / 10)):  # numero de secuencias de cada barra de progreso
            episode_return = 0
            state, _ = env.reset()
            action = agent.take_action(state)
            done = False
            while not done:
                next_state, reward, terminated, truncated, _ = env.step(action)
                done = terminated or truncated
                next_action = agent.take_action(next_state)
                episode_return += reward  
                agent.update(state,action,reward,next_state)
                state = next_state
                action = next_action
            agent.decay_epsilon()    
            return_list.append(episode_return)
            if (i_episode + 1) % 10 == 0:  # Imprimir el retorno promedio
                pbar.set_postfix({
                    'episode':
                    '%d' % (num_episodes / 10 * i + i_episode + 1),
                    'return':
                    '%.3f' % np.mean(return_list[-10:]),
                    'epsilon':
                    '%.3f' % agent.epsilon
                })
            pbar.update(1)

episodes_list = list(range(len(return_list)))
plt.plot(episodes_list, return_list)
plt.xlabel('Episodios')
plt.ylabel('Retornos')
plt.title('Q($\lambda$) en {}'.format('Acantilado'))
plt.show()
```

```python
def print_agent(agent, env, action_meaning, disaster=[], end=[]):
    for i in range(env.unwrapped.shape[0]):
        for j in range(env.env.unwrapped.shape[1]):
            if (i * env.unwrapped.shape[1] + j) in disaster:
                print('****', end=' ')
            elif (i * env.unwrapped.shape[1] + j) in end:
                print('EEEE', end=' ')
            else:
                a = agent.best_action(i * env.unwrapped.shape[1] + j)
                pi_str = ''
                for k in range(len(action_meaning)):
                    pi_str += action_meaning[k] if a[k] > 0 else 'o'
                print(pi_str, end=' ')
        print()

action_meaning = ['^', '>', 'v', '<']
print('La estrategia obtenida por el algoritmo de Q($\lambda$) es：')
print_agent(agent, env, action_meaning, list(range(37, 47)), [47])
```

```python
class SarsaLambda:
    """ Sarsa(Lambda) algorithm """
    def __init__(self, epsilon, epsilon_min, epsilon_decay, alpha, gamma, lam, n_states, n_action):
        self.Q_table = np.zeros([n_states, n_action])  # Inicializar Q(s,a) table
        self.traces = np.zeros([n_states, n_action])  # Initializar tabla de rasgos
        self.n_action = n_action  # numero de acciones
        self.alpha = alpha  # tasa de aprendizaje
        self.gamma = gamma  # factor de descuento
        self.lam = lam # descuento de rasgo
        self.epsilon = epsilon  # Parametros epsilon-greedy
        self.epsilon_min = epsilon_min
        self.epsilon_decay = epsilon_decay

    def take_action(self, state):  # epsilon-greedy seleccion de accion
        if np.random.random() < self.epsilon:
            action = np.random.randint(self.n_action)
        else:
            action = np.argmax(self.Q_table[state])
        return action

    def decay_epsilon(self):
        self.epsilon = max(self.epsilon_min, self.epsilon*self.epsilon_decay)

    def best_action(self, state):  # para imprimir la estrategia
        Q_max = np.max(self.Q_table[state])
        a = [0 for _ in range(self.n_action)]
        for i in range(self.n_action):
            if self.Q_table[state, i] == Q_max:
                a[i] = 1
        return a

    def update(self, s0, a0, r, s1, a1):
        self.traces[s0, a0] = 1
        td_error = r + self.gamma * self.Q_table[s1,a1] - self.Q_table[s0,a0]
        self.Q_table += self.alpha * td_error * self.traces
        self.traces *= self.gamma * self.lam
```

```python
env = gym.make('CliffWalking-v1')

n_states = env.observation_space.n
n_actions = env.action_space.n
np.random.seed(0)
epsilon = 1
epsilon_min = 0.01
epsilon_decay = 0.995
alpha = 0.5
gamma = 0.99
lam = 0.3
agent = SarsaLambda(epsilon, epsilon_min, epsilon_decay, alpha, gamma, lam, n_states, n_actions)
num_episodes = 2000  # Numero de secuencia que correra el agente en el ambiente

return_list = []  # Guardar el retorno de cada secuencia
for i in range(10):  # Mostrar 10 barras de progreso
    # Barra de progreso de la funcion tdqm
    with tqdm(total=int(num_episodes / 10), desc='Iteration %d' % i) as pbar:
        for i_episode in range(int(num_episodes / 10)):  # numero de secuencias de cada barra de progreso
            episode_return = 0
            state, _ = env.reset()
            action = agent.take_action(state)
            done = False
            while not done:
                next_state, reward, terminated, truncated, _ = env.step(action)
                done = terminated or truncated
                next_action = agent.take_action(next_state)
                episode_return += reward  
                agent.update(state, action, reward, next_state, next_action)
                state = next_state
                action = next_action
            agent.decay_epsilon()    
            return_list.append(episode_return)
            if (i_episode + 1) % 10 == 0:  # Imprimir el retorno promedio
                pbar.set_postfix({
                    'episode':
                    '%d' % (num_episodes / 10 * i + i_episode + 1),
                    'return':
                    '%.3f' % np.mean(return_list[-10:]),
                    'epsilon':
                    '%.3f' % agent.epsilon
                })
            pbar.update(1)

episodes_list = list(range(len(return_list)))
plt.plot(episodes_list, return_list)
plt.xlabel('Episodio')
plt.ylabel('Retorno')
plt.title('Sarsa($\lambda)$ en {}'.format('Acantilado'))
plt.show()
action_meaning = ['^', 'v', '<', '>']
print('La estrategia obtenida por Sarsa($\lambda$) es：')
print_agent(agent, env, action_meaning, list(range(37, 47)), [47])
```
