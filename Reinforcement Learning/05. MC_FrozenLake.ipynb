```python
import gymnasium as gym
import numpy as np 
import matplotlib.pyplot as plt

print(gym.__version__)
```

```python
# Crear ambiente
env = gym.make('FrozenLake-v1', desc=None, map_name="4x4", is_slippery=False, render_mode = 'human')
# Probando un episodio
observation, _ = env.reset() 
done = False
while not done:
    #Escoger una acción random
    action = env.action_space.sample()  
    #Usar la acción en el ambiente
    observation, reward, done, truncated, info = env.step(action) 

#Cerrar el ambiente    
env.close()
```

```python
action_size = env.action_space.n
state_size = env.observation_space.n
print(action_size)
print(state_size)
```

* Crea una tabla para contar las veces que ha ocurrido cada par estado-acción e inicializala con ceros.

```python

N = np.zeros([state_size, action_size])
print(N)
```

# Reto 3
## 'Política épsilon-greedy'
* Crea una función que regrese una acción random con probabilidad epsilon y la acción greedy con probabilidad 1-épsilon.

```python

def epsilon_greedy(epsilon, q_table, state):
    random_number = np.random.random()
    
    #Explotación (greedy)
    if random_number > epsilon:
        action = np.argmax(q_table[state])
    
    #Exploración (random)
    else:
        action = env.action_space.sample()
    return action
```

```python
EPSILON = 1  # Exploración vs explotación
EPISODES = 30000
```

# Reto 4
## 'Entrenamiento del agente'
* Implementa los pasos de entrenamiento del algoritmo Monte Carlo

```python

env = gym.make('FrozenLake-v1', desc=None, map_name="4x4", is_slippery=False)

for episode in range(EPISODES):
    state, _ = env.reset()
    done = False
    G = 0
    episode_history = []
    
    # Correr un episodio
    while not done:
        #Seleccionar una acción a usando la política
        #épsilon-greedy
        action = epsilon_greedy(EPSILON, Q, state)

        #Aplicar a y obtener s', r' done e info 
        next_state, reward, done, truncated, info = env.step(action)

        #Guardar la tupla de experiencia
        episode_history.append((state, action, reward))
        # Hacer s = s'
        state = next_state
    
    print(f'Episode: {episode}, final r: {reward}')

    # Aprendizaje: Actualizar Q-Table con regla MC
    for s, a, r in reversed(episode_history):
        # Actualizar G para el estado s
        G += r
        # Actualizar el número de veces que se ha visitado (s,a)
        N[s, a] += 1.0

        # Regla de actualización de MC (promedio online)
        Q[s, a] = Q[s,a] + 1/N[s,a]*(G-Q[s,a])

#Cerrar el ambiente
env.close()
```

* Crea el código necesario para que el agente entrenado juegue un episodio en al ambiente.

```python

#Obtener estado inicial
env = gym.make('FrozenLake-v1', desc=None, map_name="4x4", is_slippery=False, render_mode = 'human')
state, _ = env.reset()

for step in range(100):
    #Escoger la mejor acción de la q_table
    action = np.argmax(Q[state])
    #Obtener s', r', done, truncated e info
    next_state, reward, done, truncated, info = env.step(action)
    
    if done:
        break
    state = next_state

#Cerrar ambiente        
env.close()
```
