```python
# Librerías para tratamiento de imágenes
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Lee la imagen mediante el método de OpenCV
imagen = cv2.imread('catN5.png')
imgRGB = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)
plt.imshow(imgRGB)
plt.show()
```

```python
# Aplicando filtro de promedio
imgBlur = cv2.blur(imgRGB, ksize = (5,5))
plt.imshow(imgBlur, cmap='Accent')
plt.show()
```

```python
# Aplicando filtro gaussiano
imgGaussian = cv2.GaussianBlur(imgRGB, ksize = (5,5), sigmaX = 4)
plt.imshow(imgGaussian, cmap='Accent')
plt.show()
```

```python
# Aplicando filtro de mediana
imgMedian = cv2.medianBlur(imgRGB, ksize = 5)
plt.imshow(imgMedian, cmap='Accent')
plt.show()
```

```python
# Aplicando filtro bilateral
imgBilateral = cv2.bilateralFilter(imgRGB, 9, 75, 75)
plt.imshow(imgBilateral)
plt.show()
```

```python
# Lee la imagen mediante el método de OpenCV
img2 = cv2.imread('faceN3.png')
img2RGB = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)
plt.imshow(img2RGB)
plt.show()
```

```python
# Aplicación de filtros

img2Blur = cv2.blur(img2RGB, (5,5))
img2Gaussian = cv2.GaussianBlur(img2RGB, (5,5), sigmaX = 1)
img2Median = cv2.medianBlur(img2RGB, 5)
img2Bilateral = cv2.bilateralFilter(img2RGB, 9, 75, 75)

plt.figure(figsize=(10,10))
plt.subplot(3, 2, 1)
plt.title('Original')
plt.imshow(img2RGB)

plt.subplot(3, 2, 2)
plt.title('Promedio')
plt.imshow(img2Blur)

plt.subplot(3, 2, 3)
plt.title('Gaussiano')
plt.imshow(img2Gaussian)

plt.subplot(3, 2, 4)
plt.title('Mediana')
plt.imshow(img2Median)

plt.subplot(3, 2, 5)
plt.title('Bilateral')
plt.imshow(img2Bilateral)

plt.show()
```
