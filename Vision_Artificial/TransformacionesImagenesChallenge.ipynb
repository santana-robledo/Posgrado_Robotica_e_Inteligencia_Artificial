# Introducción al preprocesamiento de imágenes

```python
# Librerías
import cv2                        # Librería para visión artificial
import numpy as np                # Librería para operaciones entre elementos multidimensionales
import matplotlib.pyplot as plt   # Librería para graficar
```

## Adquisición de las imágenes

```python
# Lectura de una imágen
manzana= cv2.imread("manzana.jpg")
```

```python
plt.imshow(manzana)
plt.show()
```

```python
# Visualización de las imágenes
```

## Formatos de color

```python
manzanaRGB= cv2.cvtColor(manzana,cv2.COLOR_BGR2RGB)
plt.imshow(manzanaRGB,cmap="Accent")
plt.show()
```

```python
manzanaHSV = cv2.cvtColor(manzana, cv2.COLOR_BGR2HSV)
plt.imshow(manzanaHSV, cmap = 'hsv')
plt.show()
```

```python
manzanaGray = cv2.cvtColor(manzana, cv2.COLOR_BGR2GRAY)
plt.imshow(manzanaGray, cmap = 'gray')
plt.show()
```

```python
# Diferencias entre formatos de color
print(manzana[350,500])
print(manzanaRGB[350,500])
print(manzanaHSV[350,500])
print(manzanaGray[350,500])
```

## Segmentación por color

```python
# Inspección de color objetivo en la imágen
print(manzanaRGB[350,470])
print(manzanaRGB[350,570])
print(manzanaRGB[350,670])
print(manzanaRGB[500,570])
```

```python
# Segmentación de objetos a traves del color
manzanaNew = np.zeros(manzanaRGB.shape , dtype = 'int32')
for i in range(manzanaRGB.shape[0] ):
  for j in range(manzanaRGB.shape[1] ):
    if (138 <= manzanaRGB[i,j,0] <= 255) and (23 <= manzanaRGB[i,j,1] <= 104) and (0 <= manzanaRGB[i,j,2] <= 95):
      manzanaNew[i,j] = manzanaRGB[i,j]
plt.imshow(manzanaNew)
plt.show()
```
